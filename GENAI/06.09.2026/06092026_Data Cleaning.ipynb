```python
What Defines an LLM?

1. Size

LLMs are "large" because of parameter count (millions → billions → trillions).

Examples:
GPT-2 → 1.5B
GPT-3 → 175B
GPT-4 → 1.8 trillion
GPT-5 → ?
Larger size → more emergent abilities (reasoning, coding, few-shot learning).
```

```python
2. Architecture

Most modern LLMs use the Transformer (decoder-only) architecture.
Key components:

Embedding Layer → converts words/tokens into vectors.
Self-Attention → lets the model focus on relevant words in context.
Feedforward Layers → process representations.
Stacking many layers = deep contextual understanding.
```

```python
3. Training Objectives

Language Modeling Objective: Predict the next token given the context.
Example: “The cat sat on the ___” → model predicts “mat”.

Masked Language Modeling (BERT-style): Predict missing tokens.

Instruction Following (fine-tuned): Perform tasks from natural instructions.
```

```python
4. Emergent Abilities

Definition: Abilities not present in small models but that suddenly appear in larger ones.
Small models → memorize, simple tasks.
Large models → reasoning, chain-of-thought, translation, coding, abstract summarization.

Example: Only very large models show the ability to do multi-step reasoning.
```

```python
5. Multi-purpose Nature

A pre-trained LLM can:

Answer questions

Summarize documents

Translate

Generate code

Engage in dialogue

Classify text
→ All with the same base model.
```

```python
# Data Cleaning in LLM Training

Why It’s Important?

Training data often comes from web crawls (e.g., Common Crawl).
Raw data = noisy: HTML tags, duplicates, ads, broken text, emojis.

If not cleaned:

Model memorizes junk instead of patterns.
Higher risk of hallucinations, bias, inefficiency.
```

```python
Steps in Data Cleaning

1. Remove Noise
Strip HTML, JavaScript, CSS.
Remove non-text artifacts (tables, URLs if not needed).

2. Normalize
Standardize casing (all lower).
Normalize whitespace, punctuation, special characters.

3. Filter Low-Quality Text
Remove spam/gibberish.
Detect language (remove wrong language samples).

4. Toxicity Filtering
Remove hate speech, adult content (but trade-off: might reduce diversity).

5. Deduplication (often grouped with cleaning)
Detect near-duplicate or exact duplicate text.
```

```python
# dataset

import re
import html


def clean_text(text):
    """
    Remove HTML tags, backslashes, escaped sequences, and decode HTML entities.
    """
    if not isinstance(text, str):
        return text

    # Step 1: Remove all HTML/XML tags from the text
    # Input : "<h1>Welcome</h1>"
    # Output: "Welcome"
    text = re.sub(r"<[^>]+>", "", text)
    
    # Step 2: Decode HTML entities into readable characters
    # Input : "Price is &#36;100 &amp; tax"
    # Output: "Price is $100 & tax"
    text = html.unescape(text)
    
    # Step 3: Remove literal backslash characters
    # Input : "folder\\subfolder\\file"
    # Output: "folder subfolder file"
    text = re.sub(r"\\+", " ", text)
    
    # Step 4: Replace newline (\n), carriage return (\r),
    # and tab (\t) characters with spaces
    # Input : "Hello\nWorld\tGPT"
    # Output: "Hello World GPT"
    text = re.sub(r"[\n\r\t]+", " ", text)
    
    # Step 5: Normalize spacing by collapsing multiple spaces
    # Input : "Hello      World"
    # Output: "Hello World"
    text = re.sub(r"\s+", " ", text)
    
    # Step 6: Remove extra spaces from the beginning and end
    # Input : "   Hello World   "
    # Output: "Hello World"
    text = text.strip()

    return text
```

```python
# Apply the clean_text() function to the "text" column of every example
# dataset.map() processes the dataset row by row
dataset = dataset.map(
    lambda x: {"text": clean_text(x["text"])}

# Show cleaned dataset
for item in dataset:
    print(item, "\n")
```

## 1.1.2 Normalize Whitespace & Punctuation

```python
def normalize_text(text):
    text = text.lower()
    text = re.sub(r"\s+", " ", text)                          # normalize spaces
    text = text.translate(str.maketrans("", "", string.punctuation))  # remove punctuation
    return text.strip()

dataset = dataset.map(lambda x: {"text": normalize_text(x["text"])})
print("After normalization:\n", dataset[0:20]["text"])
```

## 1.1.3 Remove Gibberish / Short Sentences

```python
# Function to filter out very short texts (likely gibberish)
def filter_gibberish(example):
    if "text" not in example or not example["text"]:
        return False  # skip empty entries
    words = example["text"].split()
    return len(words) > 5  # keep texts with more than 5 words

# Apply filter
dataset = dataset.filter(filter_gibberish)

print("Dataset size after gibberish filter:", len(dataset))

# Optional: view first few cleaned entries
for i, item in enumerate(dataset):
    print(f"{i+1}. [{item['label']}] {item['text']}\n")
```
