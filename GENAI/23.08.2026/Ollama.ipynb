What is Ollama?

Ollama is a local LLM runtime that lets you run large language models on your own machine using a simple command-line or API interface.

# Think of Ollama as “Docker for LLMs”.

```python
What Problem Does Ollama Solve?

Before Ollama:
    You needed complex ML setups
    GPUs, CUDA, Python environments
    Model downloads + serving logic

With Ollama:
    One command
    Models run locally
    No cloud dependency
```

```python
# Key Characteristics

| Feature            | Explanation                         |
| ------------------ | ----------------------------------- |
| Local execution    | Models run on your laptop or server |
| Open-source models | LLaMA, Mistral, Gemma, Phi, etc.    |
| Simple interface   | CLI + REST API                      |
| Privacy-friendly   | Data never leaves your machine      |
| Cross-platform     | macOS, Linux, Windows (via WSL)     |
```

```python
How Ollama Works 

1.	You install Ollama
2.	You pull a model (like llama3, mistral)
3.	Ollama:
    o	Loads the model
    o	Handles tokenization
    o	Manages context & inference
4.	You send prompts via:
    o	Terminal
    o	Python
    o	REST API

You don’t manage the ML plumbing.
```
