```python
# Imports
from transformers import pipeline

# Load summarization pipeline (defaults to BART or T5 depending on availability)
summarizer = pipeline("summarization", model="facebook/bart-base")

# Input text
text = """
The field of artificial intelligence is growing rapidly. 
Transformers have become one of the core architectures in many NLP tasks. 
They rely heavily on attention mechanisms.
"""

# Summarize
summary = summarizer(text, max_length=50, min_length=10, do_sample=False)
print("Generated Summary:\n", summary[0]['summary_text'])
```

```python
from transformers import pipeline

generator = pipeline("text-generation", model="gpt2")

prompt = "Transformers are changing the future of AI because"
outputs = generator(prompt, max_length=40, num_return_sequences=1)

print("Generated Text:\n", outputs[0]["generated_text"])
```
