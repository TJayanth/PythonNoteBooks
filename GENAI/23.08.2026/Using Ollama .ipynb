```python
# Simple Prompt → Response
import ollama

response = ollama.chat(
    model='llama3',
    messages=[
        {
            'role': 'user',
            'content': 'Explain machine learning in one sentence.'
        }
    ]
)

print(response['message']['content'])
```

```python
# Temperature Demonstration
# the same prompt can produce different outputs.

import ollama

prompt = "Give a slogan for an AI startup."

for i in range(3):

    response = ollama.generate(
        model='llama3',
        prompt=prompt,
        options={
            'temperature': 1.0
        }
    )

    print(f"\nResponse {i+1}")
    print(response['response'])

# Low temperature → deterministic
# High temperature → creative
```

```python
# Context Awareness
messages = [
    {
        'role': 'user',
        'content': 'My name is Anish.'
    },
    {
        'role': 'assistant',
        'content': 'Nice to meet you Anish.'
    },
    {
        'role': 'user',
        'content': 'What is my name?'
    }
]

response = ollama.chat(
    model='llama3',
    messages=messages
)

print(response['message']['content'])
```

```python
# Next Token Prediction Concept

response = ollama.generate(
    model='llama3',
    prompt='The capital of France is'
)

print(response['response'])
```

```python
# Chain of Thought Example

response = ollama.generate(
    model='llama3',
    prompt='''
A store sells pencils for ₹5 each.

If I buy 8 pencils, how much do I pay?

Think step by step.
'''
)

print(response['response'])
```

```python
# Chain of Thought Example - CONVENTIONAL 

response = ollama.generate(
    model='llama3',
    prompt='''
A store sells pencils for ₹5 each.

If I buy 8 pencils, how much do I pay?

Answer only.
'''
)

print(response['response'])
```

```python
# Zero-Shot vs Few-Shot
# Zero
response = ollama.generate(
    model='llama3',
    prompt='''
Classify sentiment:

"I love this movie"
'''
)

print(response['response'])
```

```python
# Few-Shot
response = ollama.generate(
    model='llama3',
    prompt='''
Text: I love this movie
Sentiment: Positive

Text: This is terrible
Sentiment: Negative

Text: The product is amazing
Sentiment:
'''
)

print(response['response'])
```

```python
# Hallucination Example
response = ollama.generate(
    model='llama3',
    prompt='''
Who won the FIFA World Cup in 2050?
'''
)

print(response['response'])
```

```python
# Tokenization Example

from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained(
    "meta-llama/Meta-Llama-3-8B"
)

text = "Machine Learning is amazing"

tokens = tokenizer.tokenize(text)

print(tokens)
```

```python
ids = tokenizer.encode(text)

print(ids)
```

```python
# Streaming Tokens Live
import ollama

stream = ollama.generate(
    model='llama3',
    prompt='Write a short story about AI.',
    stream=True
)

for chunk in stream:
    print(chunk['response'], end='', flush=True)
```
