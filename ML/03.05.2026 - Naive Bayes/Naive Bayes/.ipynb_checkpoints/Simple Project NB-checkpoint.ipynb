```python
# Problem

Given a text ticket like:

“My laptop is not connecting to WiFi”
“Salary not credited this month”
“Cannot access email account”

Predict category:

IT Support
HR
Finance
Admin (optional)
```

```python
# System Design

1. Input
Text complaint / ticket

2. Processing
Cleaning
TF-IDF vectorization

3. Model
scikit-learn

Multinomial Naive Bayes
4. Output
Category prediction
Confidence score
```

```python
# Sample data

data = {
    "text": [
        "My laptop is not working and IT support needed",
        "Forgot password to email account",
        "Salary not credited this month",
        "Need leave approval for medical reasons",
        "System is crashing when I open software",
        "HR policy clarification required",
        "Unable to access company VPN",
        "Tax deduction issue in salary"
    ],
    "label": [
        "IT", "IT", "Finance", "HR",
        "IT", "HR", "IT", "Finance"
    ]
}
```

```python
# import pandas as pd

# df = pd.read_csv("support_tickets_10k.csv")

# print(df.head())
# print(df.shape)
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.pipeline import Pipeline
from sklearn.metrics import classification_report

df = pd.DataFrame(data)

X_train, X_test, y_train, y_test = train_test_split(
    df["text"], df["label"],
    test_size=0.25,
    random_state=42
)

model = Pipeline([
    ("tfidf", TfidfVectorizer(stop_words="english")),
    ("nb", MultinomialNB())
])

model.fit(X_train, y_train)

y_pred = model.predict(X_test)

print(classification_report(y_test, y_pred))
```

```python
# Prediction function

def predict_ticket(text):
    prediction = model.predict([text])[0]
    proba = model.predict_proba([text])[0]
    confidence = round(max(proba) * 100, 2)

    return prediction, confidence
```

```python
print(predict_ticket("My salary is not credited"))
print(predict_ticket("Laptop is overheating and slow"))
```
