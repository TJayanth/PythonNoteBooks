```python
#pip install tf-keras


from transformers import pipeline

# Sentiment Analysis
classifier = pipeline("sentiment-analysis")
print(classifier("I love teaching NLP with Hugging Face!"))

# Text Generation
generator = pipeline("text-generation", model="gpt2")
print(generator("Once upon a time", max_length=30, num_return_sequences=1))

# Named Entity Recognition (NER)
ner = pipeline("ner", grouped_entities=True)
print(ner("Hugging Face is based in New York and Paris."))
```

```python
entity_group → The type of entity recognized.

"ORG" = Organization
"LOC" = Location
score → Confidence score (probability from the model, between 0 and 1).
e.g., 0.9986 means ~99.8% confident.

word → The actual token/text from your input that was classified.
"Hugging Face", "New York", "Paris".

start / end → Character positions in the original input text.
Useful if you want to highlight or extract from the raw sentence.


Hoe to understand this ?
The model is telling you it thinks “Hugging Face” is an organization (89.6% confidence), 
“New York” is a location (99.9%), and “Paris” is also a location (99.8%).
```

```python
from transformers import AutoTokenizer, AutoModelForSequenceClassification

# Load BERT base for sentiment analysis
model_name = "distilbert-base-uncased-finetuned-sst-2-english"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSequenceClassification.from_pretrained(model_name)

# Encode text
inputs = tokenizer("Hugging Face makes NLP simple!", return_tensors="pt")

# Run model
outputs = model(**inputs)
print(outputs.logits)

```

```python
#Since 4.1337 is much larger, the model strongly predicts positive sentiment.
```

```python
import torch
from torch.nn import functional as F

logits = torch.tensor([[-3.8404, 4.1337]])
probs = F.softmax(logits, dim=1)
print(probs)
```

```python
This means:

Negative sentiment ≈ 0%
Positive sentiment ≈ 99.9%
```

# It’s basically a hub + library ecosystem for almost all modern NLP, Vision, Speech, and Multimodal tasks.
### Google Play Store for AI models or Github for ML models

```python
1. NLP (Natural Language Processing)

Tasks you can do with Hugging Face transformers:
Text Classification → sentiment analysis, topic labeling, intent detection.
Named Entity Recognition (NER) → extracting entities like people, places, organizations.
Question Answering (QA) → given a passage + question, extract the answer.
Summarization → shorten long text while preserving meaning.
Translation → English ↔ French, German, Hindi, etc.
Text Generation → story writing, code generation (e.g., using GPT-2/3-like models).
Fill-Mask → “The capital of France is <mask>” → predicts “Paris.”

2. Vision (Computer Vision with Hugging Face + 🤗 transformers)
Image Classification → predict object categories.
Object Detection → bounding boxes around objects.
Image Segmentation → pixel-wise classification (e.g., self-driving perception).
Image-to-Text (Captioning) → describe an image in natural language.

3. Speech & Audio
Automatic Speech Recognition (ASR) → speech → text (e.g., Whisper, Wav2Vec2).
Text-to-Speech (TTS) → text → speech.
Audio Classification → detect emotions, speaker ID, sound events.

4. Multimodal
CLIP models → connect text + images (e.g., “find me an image of a cat wearing sunglasses”).
Vision-Language Models (BLIP, Flamingo) → Q&A about images.
Video + Text tasks (newer models emerging).

5. Hugging Face Ecosystem Beyond Models
🤗 Datasets → 100k+ datasets ready to use.
🤗 Tokenizers → super-fast tokenization.
🤗 Spaces → host ML apps with Gradio/Streamlit.
🤗 Accelerate → easy multi-GPU / TPU training.
🤗 Diffusers → Stable Diffusion & other generative image models.
🤗 PEFT / LoRA → efficient fine-tuning of large models.
```

```python
# Text Sentiment Analysis:

from transformers import pipeline

sentiment = pipeline("sentiment-analysis")
print(sentiment("I really love teaching AI with Hugging Face!"))
```

```python
#2. Text: Named Entity Recognition (NER)

ner = pipeline("ner", grouped_entities=True)
print(ner("Hugging Face is based in New York and Paris."))
```

```python
#3. Text: Question Answering

qa = pipeline("question-answering")
print(qa({
    "question": "Where is Hugging Face based?",
    "context": "Hugging Face is a company based in New York and Paris."
}))
```

```python
#4. Vision: Image Classification

from transformers import pipeline
classifier = pipeline("image-classification", model="google/vit-base-patch16-224")
print(classifier("https://huggingface.co/datasets/huggingface/documentation-images/resolve/main/cats.png"))

```

```python
# We can directly load it

from transformers import pipeline

# Load a sentiment model fine-tuned by Hugging Face
sentiment = pipeline("sentiment-analysis", model="distilbert/distilbert-base-uncased-finetuned-sst-2-english")

print(sentiment("I love Hugging Face!"))
```

```python
## 2. Using AutoModel and AutoTokenizer
### If you want more control (fine-tuning, embeddings, etc.), you load the model + tokenizer directly:

from transformers import AutoTokenizer, AutoModelForSequenceClassification

# Load someone else’s model from Hugging Face Hub
model_name = "distilbert-base-uncased-finetuned-sst-2-english"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSequenceClassification.from_pretrained(model_name)

# Run inference manually
inputs = tokenizer("Students enjoy learning AI!", return_tensors="pt")
outputs = model(**inputs)
print(outputs.logits)   # raw predictions
```
