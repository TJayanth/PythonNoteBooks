```python
import cv2
from ultralytics import YOLO

# Load YOLOv8 pretrained CNN model
model = YOLO("yolov8n.pt")  # nano = fast for live demo

# Open webcam (0 = default camera)
cap = cv2.VideoCapture(0)

if not cap.isOpened():
    raise RuntimeError("Webcam not accessible")

while True:
    ret, frame = cap.read()
    if not ret:
        break

    # Run YOLO inference (CNN forward pass)
    results = model(frame, stream=True)

    # Draw detections
    for r in results:
        annotated_frame = r.plot()

    cv2.imshow("YOLO - Object Detection", annotated_frame)

    # Press 'q' to quit
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```

# Segmentation

```python
import cv2
from ultralytics import YOLO

# Load YOLO segmentation model
model = YOLO("yolov8n-seg.pt")

cap = cv2.VideoCapture(0)

while True:
    ret, frame = cap.read()
    if not ret:
        break

    # Run segmentation
    results = model(frame, stream=True)

    for r in results:
        annotated_frame = r.plot()  # includes masks + boxes

    cv2.imshow("YOLO Live Segmentation", annotated_frame)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```

Input: Images from conveyor-belt cameras or handheld devices
CNN Task: Image classification / object detection

    Torn packaging

    Damage probability score
