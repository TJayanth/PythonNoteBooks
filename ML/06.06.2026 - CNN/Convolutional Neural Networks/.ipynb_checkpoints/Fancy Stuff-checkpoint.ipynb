```python
import torch
import torch.nn.functional as F
from torchvision import transforms
import clip
import matplotlib.pyplot as plt
from PIL import Image
import numpy as np
from tqdm import tqdm

# Load model
device = "cuda" if torch.cuda.is_available() else "cpu"
model, preprocess = clip.load("ViT-B/32", device=device)

# Target prompt
prompt = "A tiger eating pizza"
text_token = clip.tokenize([prompt]).to(device)

# Initialize a random image (normalized between 0 and 1)
img_size = 224
image = torch.randn(1, 3, img_size, img_size, requires_grad=True, device=device)

# Optimizer
optimizer = torch.optim.Adam([image], lr=0.05)

# Transform to match CLIP expectations
normalize = transforms.Normalize(mean=[0.48145466, 0.4578275, 0.40821073],
                                 std=[0.26862954, 0.26130258, 0.27577711])

# Optimization loop
steps = 300
losses = []

for step in tqdm(range(steps)):
    optimizer.zero_grad()
    
    # Clip expects normalized images
    image_clip = normalize(image.squeeze(0)).unsqueeze(0)
    image_clip = F.interpolate(image_clip, size=(224, 224))

    # Encode image and text
    image_features = model.encode_image(image_clip)
    text_features = model.encode_text(text_token)

    # Cosine similarity as objective (maximize similarity)
    image_features = image_features / image_features.norm(dim=-1, keepdim=True)
    text_features = text_features / text_features.norm(dim=-1, keepdim=True)
    similarity = (image_features @ text_features.T).squeeze()

    loss = -similarity
    loss.backward()
    optimizer.step()

    # Clamp pixel values to [0,1]
    with torch.no_grad():
        image.clamp_(0, 1)

    losses.append(loss.item())

    # Print every 50 steps
    if step % 50 == 0 or step == steps - 1:
        print(f"Step {step} | Loss: {loss.item():.4f}")

# Show the image
final_img = image.detach().cpu().squeeze(0).permute(1, 2, 0).numpy()
plt.imshow(final_img)
plt.title(f"Generated: {prompt}")
plt.axis("off")
plt.show()
```

```python
import cv2

# === Choose tracker: CSRT or KCF (KCF is faster, CSRT is more accurate) ===
TRACKER_TYPE = "CSRT"

# OpenCV tracker dictionary
OPENCV_OBJECT_TRACKERS = {
    "KCF": cv2.legacy.TrackerKCF_create,
    "CSRT": cv2.legacy.TrackerCSRT_create
}

# Open webcam
cap = cv2.VideoCapture(0)  # 0 for default webcam
assert cap.isOpened(), "Could not open webcam"

# Read first frame
ok, frame = cap.read()
if not ok:
    print("Cannot read webcam frame")
    exit()

# Let user select the object to track (draw ROI)
bbox = cv2.selectROI("Select object to track", frame, fromCenter=False, showCrosshair=True)
cv2.destroyWindow("Select object to track")

# Initialize tracker
tracker = OPENCV_OBJECT_TRACKERS[TRACKER_TYPE]()
tracker.init(frame, bbox)

# Start tracking loop
while True:
    ok, frame = cap.read()
    if not ok:
        break

    # Update tracker
    ok, box = tracker.update(frame)

    # Draw bounding box
    if ok:
        x, y, w, h = [int(v) for v in box]
        cv2.rectangle(frame, (x, y), (x+w, y+h), (0,255,0), 2)
        cv2.putText(frame, TRACKER_TYPE + " Tracker", (x, y-10),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0,255,0), 2)
    else:
        cv2.putText(frame, "Tracking failure", (50,80),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.75, (0,0,255), 2)

    cv2.imshow("Live Object Tracking", frame)

    # Press ESC to quit
    if cv2.waitKey(1) & 0xFF == 27:
        break

cap.release()
cv2.destroyAllWindows()
```
