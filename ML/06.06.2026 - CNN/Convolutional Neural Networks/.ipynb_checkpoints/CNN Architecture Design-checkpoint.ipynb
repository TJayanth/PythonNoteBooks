# "Lets design and train a CNN from scratch to classify images"

```python
class MyFirstCNN(nn.Module):
    def __init__(self):
        super().__init__()
        self.conv1 = nn.Conv2d(1, 8, 3)  # (28 -> 26)
        self.pool = nn.MaxPool2d(2, 2)   # (26 -> 13)
        self.fc1 = nn.Linear(8 * 13 * 13, 10)

    def forward(self, x):
        x = F.relu(self.pool(self.conv1(x)))
        x = x.view(-1, 8 * 13 * 13)
        x = self.fc1(x)
        return x
```

```python
1.
self.conv1 = nn.Conv2d(1, 8, 3)
    1: Input channels ( grayscale images, 1 channel).
    8: Output channels → this creates 8 filters, so the output will have 8 feature maps.
    3: Kernel size = 3x3

Size Calculation (no padding):
    If input = 28×28, kernel=3 → output = (28 − 3 + 1) = 26×26

So:
[1, 28, 28] → conv → [8, 26, 26]


2.
self.pool = nn.MaxPool2d(2, 2) 
    Applies a 2x2 max pooling window.
    Stride = 2 → so it reduces spatial size by half.
So:
    [8, 26, 26] → pooling → [8, 13, 13]

3.
self.fc1 = nn.Linear(8 * 13 * 13, 10)
    Input: 8 feature maps of size 13×13 = 8 × 13 × 13 = 1352 inputs
    Output: 10 classes (e.g., digits 0–9)

This FC layer combines all the learned features to predict the final class



# x = F.relu(self.pool(self.conv1(x)))
    self.conv1(x) applies 8 filters → shape becomes [8, 26, 26]
    self.pool(...) downsamples → [8, 13, 13]
    F.relu(...) adds non-linearity


# x = x.view(-1, 8 * 13 * 13)
    Converts 3D tensor [8, 13, 13] into a flat 1D vector of size 1352
    -1 lets PyTorch automatically infer the batch size

Now the data is ready to be passed into the dense (fully connected) layer.


```

```python
class MyCIFARCNN(nn.Module): # 10 class
    def __init__(self):
        super().__init__()
        self.conv1 = nn.Conv2d(3, 16, 3, padding=1)
        self.conv2 = nn.Conv2d(16, 32, 3, padding=1)
        self.pool = nn.MaxPool2d(2, 2)
        self.fc1 = nn.Linear(32 * 8 * 8, 64)
        self.fc2 = nn.Linear(64, 10)

    def forward(self, x):
        x = F.relu(self.pool(self.conv1(x))) 
        x = F.relu(self.pool(self.conv2(x)))
        x = x.view(-1, 32 * 8 * 8)
        x = F.relu(self.fc1(x))
        x = self.fc2(x)
        return x
```

```python
1. self.conv1 = nn.Conv2d(3, 16, 3, padding=1)
    3: Input channels (for RGB images, which have 3 color channels)
    16: Number of filters → output will have 16 feature maps
    3: Kernel size = 3×3
    padding=1: This ensures that the output retains the same width and height as the input (32×32)

Input:

[3, 32, 32] → Conv2d(3, 16, 3, padding=1) → [16, 32, 32]



2. self.pool = nn.MaxPool2d(2, 2)

    2×2 pooling window, stride = 2
    Reduces each spatial dimension by half

After pooling:
[16, 32, 32] → [16, 16, 16]


3. self.conv2 = nn.Conv2d(16, 32, 3, padding=1)

    Takes 16 input channels and produces 32 output channels
    Same 3×3 kernel with padding=1
Input to this layer:
[16, 16, 16] → Conv2d(16, 32, 3, padding=1) → [32, 16, 16]

Then again passed through max pooling:
[32, 16, 16] → MaxPool2d → [32, 8, 8]


4.self.fc1 = nn.Linear(32 * 8 * 8, 64)

    Input is the flattened tensor from the convolutional layers: 32 × 8 × 8 = 2048
    Output: 64 neurons
    Acts as a hidden layer (introduces more learnable weights)

5. self.fc2 = nn.Linear(64, 10)

    Final classification layer
    Outputs 10 class scores 


def forward(self, x):
    x = F.relu(self.pool(self.conv1(x)))  # (32×32 → 16×16)
    x = F.relu(self.pool(self.conv2(x)))  # (16×16 → 8×8)
    x = x.view(-1, 32 * 8 * 8)            # Flatten → [2048]
    x = F.relu(self.fc1(x))               # Dense → [64]
    x = self.fc2(x)                       # Output → [10]
    return x

```

```python
| Layer   | Output Shape   | Explanation                          |
| ------- | -------------- | ------------------------------------ |
| Input   | `[3, 32, 32]`  | RGB image                            |
| Conv1   | `[16, 32, 32]` | 16 filters, padding keeps size same  |
| Pool1   | `[16, 16, 16]` | Downsampled by 2                     |
| Conv2   | `[32, 16, 16]` | 32 filters                           |
| Pool2   | `[32, 8, 8]`   | Downsampled again                    |
| Flatten | `[2048]`       | 32×8×8 = 2048                        |
| FC1     | `[64]`         | Hidden dense layer                   |
| FC2     | `[10]`         | Class scores (logits for 10 classes) |
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F

class CNN3Layer(nn.Module):
    def __init__(self):
        super(CNN3Layer, self).__init__()
        # Convolution layers
        self.conv1 = nn.Conv2d(in_channels=3, out_channels=16, kernel_size=3, padding=1)  # [3,32,32] -> [16,32,32]
        self.conv2 = nn.Conv2d(16, 32, 3, padding=1)                                      # [16,16,16] -> [32,16,16]
        self.conv3 = nn.Conv2d(32, 64, 3, padding=1)                                      # [32,8,8] -> [64,8,8]
        self.pool = nn.MaxPool2d(2, 2)

        # Fully connected layers
        self.fc1 = nn.Linear(64 * 4 * 4, 128)  # After 3 poolings: 32 → 16 → 8 → 4
        self.fc2 = nn.Linear(128, 10)          # 10 classes

    def forward(self, x):
        x = self.pool(F.relu(self.conv1(x)))   # -> [16, 16, 16]
        x = self.pool(F.relu(self.conv2(x)))   # -> [32, 8, 8]
        x = self.pool(F.relu(self.conv3(x)))   # -> [64, 4, 4]

        x = x.view(-1, 64 * 4 * 4)             # Flatten to 1024 features
        x = F.relu(self.fc1(x))                # Fully connected with ReLU
        x = self.fc2(x)                        # Final output (logits)
        return x
```

```python
from torchvision.datasets import Flowers102
from torchvision import transforms
from torch.utils.data import DataLoader
import torch

# Define transforms: resize + convert to tensor + normalize
transform = transforms.Compose([
    transforms.Resize((32, 32)),     # Resize to fit your CNN input
    transforms.ToTensor(),
    transforms.Normalize([0.5]*3, [0.5]*3)  # Normalize for RGB (3 channels)
])

# Load dataset
train_data = Flowers102(root='data', split='train', transform=transform, download=False)
val_data = Flowers102(root='data', split='val', transform=transform, download=False)

# Dataloaders
train_loader = DataLoader(train_data, batch_size=32, shuffle=True)
val_loader = DataLoader(val_data, batch_size=32)
```

```python
import torch.nn as nn
import torch.nn.functional as F

class CNN3LayerFlowers(nn.Module):
    def __init__(self):
        super().__init__()
        self.conv1 = nn.Conv2d(3, 16, 3, padding=1)
        self.conv2 = nn.Conv2d(16, 32, 3, padding=1)
        self.conv3 = nn.Conv2d(32, 64, 3, padding=1)
        self.pool = nn.MaxPool2d(2, 2)

        self.fc1 = nn.Linear(64 * 4 * 4, 128)
        self.fc2 = nn.Linear(128, 102)  # 102 flower classes

    def forward(self, x):
        x = self.pool(F.relu(self.conv1(x)))  # 32x32 -> 16x16
        x = self.pool(F.relu(self.conv2(x)))  # 16x16 -> 8x8
        x = self.pool(F.relu(self.conv3(x)))  # 8x8 -> 4x4
        x = x.view(-1, 64 * 4 * 4)
        x = F.relu(self.fc1(x))
        x = self.fc2(x)
        return x
```

```python
model = CNN3LayerFlowers()
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
model = model.to(device)

optimizer = torch.optim.Adam(model.parameters(), lr=0.001)
loss_fn = nn.CrossEntropyLoss()

for epoch in range(5):  # Try more epochs later
    model.train()
    for images, labels in train_loader:
        images, labels = images.to(device), labels.to(device)

        optimizer.zero_grad()
        outputs = model(images)
        loss = loss_fn(outputs, labels)
        loss.backward()
        optimizer.step()

    print(f"Epoch {epoch+1} complete")
```

```python
from sklearn.metrics import confusion_matrix, classification_report, ConfusionMatrixDisplay
import matplotlib.pyplot as plt
import torch
import numpy as np
```

```python
model.eval()
all_preds = []
all_labels = []

with torch.no_grad():
    for images, labels in val_loader:
        images = images.to(device)
        outputs = model(images)
        _, preds = torch.max(outputs, 1)

        all_preds.extend(preds.cpu().numpy())
        all_labels.extend(labels.numpy())
```

```python
cm = confusion_matrix(all_labels, all_preds)
print("Confusion Matrix Shape:", cm.shape)

disp = ConfusionMatrixDisplay(confusion_matrix=cm)
fig, ax = plt.subplots(figsize=(20, 20))  # You can make this larger
disp.plot(ax=ax)
plt.title("Confusion Matrix - Flowers102")
plt.show()


from torchvision.datasets import Flowers102

_, class_to_idx = Flowers102(root='data', split='train', download=False).class_to_idx.items()
idx_to_class = {v: k for k, v in class_to_idx}
```

```python
print("Classification Report:")
print(classification_report(all_labels, all_preds, digits=3))
```
