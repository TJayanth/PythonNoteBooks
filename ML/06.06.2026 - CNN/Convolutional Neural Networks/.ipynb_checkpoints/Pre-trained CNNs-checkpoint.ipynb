# Leveraging Pre-Trained CNNs for Image Tasks

```python
Introduction to Transfer Learning

    What is transfer learning?
        Using a model trained on one task as the starting point for a related task.

    Why use pre-trained CNNs?
        Saves time and compute.
        Useful when labeled data is scarce.
        Often yields better performance than training from scratch.
```

```python
Overview of Popular Pre-Trained CNN Architectures

    AlexNet (ImageNet challenge breakthrough, 2012)
    VGGNet (Very Deep CNNs)
    ResNet (Residual Connections – deeper without vanishing gradients)
    Inception (GoogLeNet) (Multi-scale feature extraction)
    MobileNet / EfficientNet (Lightweight, mobile-optimized)
```

```python
| Architecture                 | Year | Depth (Layers) | Parameters (Millions) | Top-1 Accuracy (ImageNet) | Key Feature                                    |
| ---------------------------- | ---- | -------------- | --------------------- | ------------------------- | ---------------------------------------------- |
| **AlexNet**                  | 2012 | 8              | \~60M                 | \~57%                     | First deep CNN to win ImageNet (ReLU, dropout) |
| **VGG-16**                   | 2014 | 16             | \~138M                | \~71.5%                   | Deep with uniform 3x3 conv filters             |
| **GoogLeNet (Inception v1)** | 2014 | 22             | \~6.8M                | \~69.8%                   | Inception modules for multi-scale filtering    |
| **ResNet-50**                | 2015 | 50             | \~25.6M               | \~76.2%                   | Residual (skip) connections                    |
| **DenseNet-121**             | 2017 | 121            | \~8M                  | \~74.9%                   | Dense connections (feature reuse)              |
| **MobileNet v2**             | 2018 | \~53           | \~3.4M                | \~71.8%                   | Lightweight, optimized for mobile              |
| **EfficientNet-B0**          | 2019 | \~82           | \~5.3M                | \~77.1%                   | Compound scaling (depth, width, res.)          |
```

```python
Top-1 Accuracy: Percentage of times the top prediction is correct (on ImageNet validation set).
Parameters: Lower is better for mobile/edge devices, higher may imply more capacity but also more compute.

Use Case Suggestions:
    MobileNet / EfficientNet-B0: Use in mobile/IoT or embedded projects.
    ResNet-50: Balanced trade-off; ideal for fine-tuning.
    VGG-16: Simple and interpretable, but heavy.
```

# Feature Extraction vs Fine-Tuning in Pre-Trained CNNs

```python
| Aspect                  | **Feature Extraction**                            | **Fine-Tuning**                                           |
| ----------------------- | ------------------------------------------------- | --------------------------------------------------------- |
| **Layer Freezing**      | All pre-trained layers are frozen                 | Some top layers of the pre-trained model are unfrozen     |
| **Training Speed**      | Faster (fewer parameters to train)                | Slower (more parameters updated)                          |
| **Risk of Overfitting** | Lower (small model capacity for new data)         | Higher (especially with small datasets)                   |
| **Learning Capacity**   | Limited to what the pre-trained model has learned | Greater flexibility to adapt to new tasks                 |
| **Use Case**            | When data is limited or very similar to ImageNet  | When more data is available or domain is quite different  |
| **Typical Code**        | `base_model.trainable = False`                    | `base_model.trainable = True` *(or selectively unfreeze)* |
```

```python
# A: Feature Extraction

import torch
import torch.nn as nn
import torchvision.models as models

# Load pre-trained ResNet50 model
base_model = models.resnet50(pretrained=True)

# Freeze all layers (feature extraction mode)
for param in base_model.parameters():
    param.requires_grad = False

# Replace the classifier (fully connected layer)
# Original ResNet50 fc layer: (in_features=2048, out_features=1000)
base_model.fc = nn.Sequential(
    nn.Linear(2048, 128),
    nn.ReLU(),
    nn.Linear(128, 10),          # For 10 classes
    nn.Softmax(dim=1)            # Optional: include softmax if using cross-entropy with logits=False
)

# Move to GPU if available
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = base_model.to(device)

# Print model architecture (optional)
print(model)

```

```python
# Fine tuning

import torch
import torch.nn as nn
import torchvision.models as models

# Load pre-trained ResNet50
base_model = models.resnet50(pretrained=True)

# Freeze all layers initially
for param in base_model.parameters():
    param.requires_grad = False

# Unfreeze last few layers (e.g., layer4 block + fc)
for name, param in base_model.named_parameters():
    if "layer4" in name or "fc" in name:
        param.requires_grad = True

# Replace the classification head (fc layer)
base_model.fc = nn.Sequential(
    nn.Linear(2048, 128),
    nn.ReLU(),
    nn.Linear(128, 10),       # For 10-class problem
    nn.Softmax(dim=1)         # Optional (skip if using CrossEntropyLoss)
)

# Move model to device
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = base_model.to(device)

# Print trainable parameters (for verification)
print("Trainable layers:")
for name, param in model.named_parameters():
    if param.requires_grad:
        print(name)

```

```python
# Feature extraction

import torch
import torchvision.models as models
import torchvision.transforms as transforms
from PIL import Image
import matplotlib.pyplot as plt

# Load pre-trained ResNet18
model = models.resnet18(pretrained=True)
model.eval()  # Set to eval mode
feature_extractor = torch.nn.Sequential(*list(model.children())[:-1])  # Remove the final FC layer

# Image preprocessing (match ImageNet)
transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], 
                         std=[0.229, 0.224, 0.225])
])


img_path = "redpanda.jpg"
img = Image.open(img_path).convert('RGB')
img_tensor = transform(img).unsqueeze(0)  # Add batch dimension

# Extract features
with torch.no_grad():
    features = feature_extractor(img_tensor)

# Flatten features and show result
features = features.view(features.size(0), -1)  # Shape: [1, 512]
print("Feature vector shape:", features.shape)

# Show image
plt.imshow(img)
plt.title(f"Extracted Feature Shape: {features.shape}")
plt.axis('off')
plt.show()
```

```python
import torch
import torchvision.models as models
import torchvision.transforms as transforms
from PIL import Image
import matplotlib.pyplot as plt
import numpy as np
import cv2

# Load pretrained ResNet18
model = models.resnet18(pretrained=True)
model.eval()

# Choose target layer (last conv layer)
target_layer = model.layer4[1].conv2  # You can experiment with conv1, layer1[0], etc.

# Register hooks to get gradients and activations
gradients = []
activations = []

def backward_hook(module, grad_input, grad_output):
    gradients.append(grad_output[0])

def forward_hook(module, input, output):
    activations.append(output)

# Register hooks
target_layer.register_forward_hook(forward_hook)
target_layer.register_backward_hook(backward_hook)

# Preprocess image
transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406],
                         std=[0.229, 0.224, 0.225])
])

img_path = "redpanda.jpg"
img = Image.open(img_path).convert('RGB')
img_tensor = transform(img).unsqueeze(0)

# Forward pass
output = model(img_tensor)
pred_class = output.argmax().item()

# Backward pass
model.zero_grad()
class_loss = output[0, pred_class]
class_loss.backward()

# Grad-CAM computation
grad = gradients[0].squeeze().cpu().detach().numpy()        # [C, H, W]
act = activations[0].squeeze().cpu().detach().numpy()       # [C, H, W]

weights = np.mean(grad, axis=(1, 2))                        # [C]
cam = np.sum(weights[:, None, None] * act, axis=0)          # [H, W]
cam = np.maximum(cam, 0)  # ReLU

# Normalize CAM
cam = cam - np.min(cam)
cam = cam / np.max(cam)
cam = cv2.resize(cam, (224, 224))

# Visualize
img_original = np.array(img.resize((224, 224))) / 255.0
heatmap = cv2.applyColorMap(np.uint8(255 * cam), cv2.COLORMAP_JET)
heatmap = np.float32(heatmap) / 255
overlayed = heatmap + img_original
overlayed = overlayed / overlayed.max()

plt.figure(figsize=(10, 4))
plt.subplot(1, 3, 1)
plt.imshow(img_original)
plt.title("Original Image")
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(cam, cmap='jet')
plt.title("Grad-CAM Heatmap")
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(overlayed)
plt.title("Overlay")
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
import torch
import torchvision.models as models
import torchvision.transforms as transforms
from PIL import Image
import matplotlib.pyplot as plt
import numpy as np
import cv2

# Load pre-trained ResNet18
model = models.resnet18(pretrained=True)
model.eval()

# Image transform
transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406], 
        std=[0.229, 0.224, 0.225])
])

# Load and preprocess image
img_path = "elephant.jpg"
img = Image.open(img_path).convert('RGB')
img_tensor = transform(img).unsqueeze(0)

# Layers to visualize
target_layers = {
    "conv1": model.conv1,
    "layer1[0].conv2": model.layer1[0].conv2,
    "layer4[1].conv2": model.layer4[1].conv2,
}

results = []

for name, layer in target_layers.items():
    gradients = []
    activations = []

    # Register hooks
    def backward_hook(module, grad_input, grad_output):
        gradients.append(grad_output[0])

    def forward_hook(module, input, output):
        activations.append(output)

    h1 = layer.register_forward_hook(forward_hook)
    h2 = layer.register_backward_hook(backward_hook)

    # Forward + backward pass
    output = model(img_tensor)
    pred_class = output.argmax().item()
    model.zero_grad()
    output[0, pred_class].backward()

    # Get activations and gradients
    grad = gradients[0].squeeze().cpu().detach().numpy()
    act = activations[0].squeeze().cpu().detach().numpy()
    weights = np.mean(grad, axis=(1, 2))
    cam = np.sum(weights[:, None, None] * act, axis=0)
    cam = np.maximum(cam, 0)
    cam = (cam - cam.min()) / (cam.max() - cam.min() + 1e-6)
    cam_resized = cv2.resize(cam, (224, 224))

    # Create overlay
    img_np = np.array(img.resize((224, 224))) / 255.0
    heatmap = cv2.applyColorMap(np.uint8(255 * cam_resized), cv2.COLORMAP_JET)
    heatmap = np.float32(heatmap) / 255
    overlay = heatmap + img_np
    overlay = overlay / np.max(overlay)

    results.append((name, cam_resized, overlay))

    # Remove hooks
    h1.remove()
    h2.remove()

# Plot everything
fig, axs = plt.subplots(len(results), 3, figsize=(10, 3 * len(results)))
if len(results) == 1:
    axs = [axs]  # Ensure it's iterable

for idx, (name, heat, overlay) in enumerate(results):
    axs[idx][0].imshow(img)
    axs[idx][0].set_title(f"{name} - Original")
    axs[idx][0].axis('off')

    axs[idx][1].imshow(heat, cmap='jet')
    axs[idx][1].set_title("Grad-CAM Heatmap")
    axs[idx][1].axis('off')

    axs[idx][2].imshow(overlay)
    axs[idx][2].set_title("Overlay")
    axs[idx][2].axis('off')

plt.tight_layout()
plt.show()
```

```python
import torch
import torchvision.models as models
import torchvision.transforms as transforms
from PIL import Image
import matplotlib.pyplot as plt
import matplotlib.animation as animation
import numpy as np

# Load model
model = models.resnet18(pretrained=True).eval()

# Transform and image
transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225])
])
img = Image.open("redpanda.jpg").convert("RGB")
input_tensor = transform(img).unsqueeze(0)

# Target layers
layers_to_hook = {
    'conv1': model.conv1,
    'layer1[0].conv2': model.layer1[0].conv2,
    'layer4[1].conv2': model.layer4[1].conv2
}

feature_maps = {}

# Hook to capture outputs
def hook_fn(name):
    def fn(module, input, output):
        feature_maps[name] = output.detach()
    return fn

# Register hooks
hooks = []
for name, layer in layers_to_hook.items():
    hooks.append(layer.register_forward_hook(hook_fn(name)))

# Forward pass
_ = model(input_tensor)

# Remove hooks
for h in hooks:
    h.remove()

# Create animation of feature maps for each layer
for name, fmap in feature_maps.items():
    fmap = fmap.squeeze(0)  # [C, H, W]
    fmap = fmap[:16]  # show first 16 filters only
    num_filters = fmap.size(0)
    cols = 4
    rows = int(np.ceil(num_filters / cols))

    fig, axs = plt.subplots(rows, cols, figsize=(8, 2 * rows))
    axs = axs.flatten()

    ims = []
    for i in range(num_filters):
        axs[i].axis('off')
        im = axs[i].imshow(fmap[i].cpu().numpy(), cmap='viridis', animated=True)
        axs[i].set_title(f"Filter {i}")
        ims.append([im])

    ani = animation.ArtistAnimation(fig, ims, interval=500, repeat_delay=1000)
    ani.save(f"{name}_filters.gif", writer='pillow')
    plt.close()

print("🎥 Saved filter activations as GIFs for conv1, layer1, and layer4.")
```

# Fine tuning

```python
import torch
import torch.nn as nn
import torch.optim as optim
from torchvision import datasets, models, transforms
from torch.utils.data import DataLoader
import time

# Paths
train_dir = 'finetune/train'
val_dir = 'finetune/val'

# Device
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# Transforms
data_transforms = {
    'train': transforms.Compose([
        transforms.Resize((224, 224)),
        transforms.RandomHorizontalFlip(),
        transforms.ToTensor(),
        transforms.Normalize([0.485, 0.456, 0.406],
                             [0.229, 0.224, 0.225])
    ]),
    'val': transforms.Compose([
        transforms.Resize((224, 224)),
        transforms.ToTensor(),
        transforms.Normalize([0.485, 0.456, 0.406],
                             [0.229, 0.224, 0.225])
    ]),
}

# Datasets and loaders
image_datasets = {
    'train': datasets.ImageFolder(train_dir, transform=data_transforms['train']),
    'val': datasets.ImageFolder(val_dir, transform=data_transforms['val'])
}
dataloaders = {
    x: DataLoader(image_datasets[x], batch_size=32, shuffle=True, num_workers=2)
    for x in ['train', 'val']
}

# Load ResNet18
model = models.resnet18(pretrained=True)

# Replace final FC layer to match number of classes
num_classes = len(image_datasets['train'].classes)
model.fc = nn.Linear(model.fc.in_features, num_classes)
model = model.to(device)

#Optional: freeze early layers (for feature extraction)
# for param in model.parameters():
#     param.requires_grad = False
# for param in model.fc.parameters():
#     param.requires_grad = True

# Loss and optimizer
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=1e-4)

# Training loop
num_epochs = 3
for epoch in range(num_epochs):
    print(f"Epoch {epoch+1}/{num_epochs}")
    for phase in ['train', 'val']:
        if phase == 'train':
            model.train()
        else:
            model.eval()

        running_loss = 0.0
        running_corrects = 0

        for inputs, labels in dataloaders[phase]:
            inputs, labels = inputs.to(device), labels.to(device)

            optimizer.zero_grad()

            with torch.set_grad_enabled(phase == 'train'):
                outputs = model(inputs)
                _, preds = torch.max(outputs, 1)
                loss = criterion(outputs, labels)

                if phase == 'train':
                    loss.backward()
                    optimizer.step()

            running_loss += loss.item() * inputs.size(0)
            running_corrects += torch.sum(preds == labels.data)

        epoch_loss = running_loss / len(image_datasets[phase])
        epoch_acc = running_corrects.double() / len(image_datasets[phase])

        print(f"{phase.capitalize()} Loss: {epoch_loss:.4f} Acc: {epoch_acc:.4f}")

print("✅ Training complete")
```

```python
# ASSIGNEMENT

import torch
from torchvision import datasets, transforms
from torch.utils.data import DataLoader, random_split

# Paths
data_dir = 'finetune/Assignment'

# Device
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# Transforms
train_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.RandomHorizontalFlip(),
    transforms.ToTensor(),
    transforms.Normalize([0.485, 0.456, 0.406],
                         [0.229, 0.224, 0.225])
])

val_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize([0.485, 0.456, 0.406],
                         [0.229, 0.224, 0.225])
])

# Load full dataset
full_dataset = datasets.ImageFolder(
    root=data_dir,
    transform=train_transform   # temp transform
)

print("Total images:", len(full_dataset))
print("Classes:", full_dataset.classes)

# Split sizes
train_ratio = 0.8
train_size = int(train_ratio * len(full_dataset))
val_size = len(full_dataset) - train_size

# Reproducible split
generator = torch.Generator().manual_seed(42)
train_dataset, val_dataset = random_split(
    full_dataset, [train_size, val_size], generator=generator
)

# Assign correct transforms
train_dataset.dataset.transform = train_transform
val_dataset.dataset.transform = val_transform

# DataLoaders
dataloaders = {
    'train': DataLoader(train_dataset, batch_size=32, shuffle=True, num_workers=2),
    'val': DataLoader(val_dataset, batch_size=32, shuffle=False, num_workers=2)
}

print("Train images:", len(train_dataset))
print("Validation images:", len(val_dataset))


# Load ResNet18
model = models.resnet18(pretrained=True)

# Replace final FC layer to match number of classes
num_classes = len(image_datasets['train'].classes)
model.fc = nn.Linear(model.fc.in_features, num_classes)
model = model.to(device)

#Optional: freeze early layers (for feature extraction)
# for param in model.parameters():
#     param.requires_grad = False
# for param in model.fc.parameters():
#     param.requires_grad = True

# Loss and optimizer
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=1e-4)

# Training loop
num_epochs = 1
for epoch in range(num_epochs):
    print(f"Epoch {epoch+1}/{num_epochs}")
    for phase in ['train', 'val']:
        if phase == 'train':
            model.train()
        else:
            model.eval()

        running_loss = 0.0
        running_corrects = 0

        for inputs, labels in dataloaders[phase]:
            inputs, labels = inputs.to(device), labels.to(device)

            optimizer.zero_grad()

            with torch.set_grad_enabled(phase == 'train'):
                outputs = model(inputs)
                _, preds = torch.max(outputs, 1)
                loss = criterion(outputs, labels)

                if phase == 'train':
                    loss.backward()
                    optimizer.step()

            running_loss += loss.item() * inputs.size(0)
            running_corrects += torch.sum(preds == labels.data)

        epoch_loss = running_loss / len(image_datasets[phase])
        epoch_acc = running_corrects.double() / len(image_datasets[phase])

        print(f"{phase.capitalize()} Loss: {epoch_loss:.4f} Acc: {epoch_acc:.4f}")

print("✅ Training complete")
```

```python
# function for a unknown image

def predict_image(image_path, model, transform, class_names):
    image = Image.open(image_path).convert("RGB")
    image = transform(image).unsqueeze(0)  # add batch dim
    image = image.to(device)

    with torch.no_grad():
        outputs = model(image)
        _, pred = torch.max(outputs, 1)

    return class_names[pred.item()]
```
