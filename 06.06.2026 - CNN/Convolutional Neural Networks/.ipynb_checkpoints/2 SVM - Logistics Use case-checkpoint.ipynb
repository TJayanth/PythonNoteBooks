```python
# Last mile dataset

## Delivery Delay Classification
```

```python
import pandas as pd
import numpy as np
from pathlib import Path

from sklearn.svm import SVC
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.model_selection import StratifiedKFold, cross_val_score

import matplotlib.pyplot as plt
```

```python
# Load Pickup & Delivery Data
# Define paths
pickup_files = Path("raw\pickup").glob("*.csv")
delivery_files = Path("raw\delivery").glob("*.csv")

# Read all CSVs
pickup_df = pd.concat([pd.read_csv(f) for f in pickup_files], ignore_index=True)
delivery_df = pd.concat([pd.read_csv(f) for f in delivery_files], ignore_index=True)

print("Pickup data shape:", pickup_df.shape)
print("Delivery data shape:", delivery_df.shape)
```

```python
# Merge Pickup & Delivery Data

# Merge on order_id (adjust if your key column is different)
df = pickup_df.merge(delivery_df, on="order_id", how="inner")
print("Merged dataset shape:", df.shape)
df.head()
```

```python
# Add dummy year to avoid OutOfBoundsDatetime
df["pickup_time"] = pd.to_datetime(
    "2023-" + df["pickup_time"].astype(str),
    format="%Y-%m-%d %H:%M:%S",
    errors="coerce"
)
df["delivery_time"] = pd.to_datetime(
    "2023-" + df["delivery_time"].astype(str),
    format="%Y-%m-%d %H:%M:%S",
    errors="coerce"
)

# Drop any rows where parsing failed
df.dropna(subset=["pickup_time", "delivery_time"], inplace=True)
```

```python
# Feature Engineering

# Delivery duration in minutes
df["delivery_duration_min"] = (df["delivery_time"] - df["pickup_time"]).dt.total_seconds() / 60

# Time-based features
df["pickup_hour"] = df["pickup_time"].dt.hour
df["pickup_dayofweek"] = df["pickup_time"].dt.dayofweek

# Optional: ensure 'distance' column exists, else create dummy
if "distance" not in df.columns:
    df["distance"] = np.random.randint(1, 20, size=len(df))  # dummy distance for demo

df.head()
```

```python
# Define Features (X) and Target (y)
# Features
X = df[["pickup_hour", "pickup_dayofweek", "distance"]]

# Binary target: delayed if > 75th percentile
threshold = df["delivery_duration_min"].quantile(0.75)
df["delayed"] = (df["delivery_duration_min"] > threshold).astype(int)
y = df["delayed"]

print(X.shape, y.shape)
print(y.value_counts())
```

```python
# 10-Fold Cross-Validation Setup

cv = StratifiedKFold(n_splits=10, shuffle=True, random_state=42)
```

```python
# Linear SVM Pipeline
linear_pipeline = Pipeline([
    ("scaler", StandardScaler()),
    ("svm", SVC(kernel="linear", C=1.0))
])

linear_scores = cross_val_score(
    linear_pipeline,
    X,
    y,
    cv=cv,
    scoring="accuracy"
)

for i, score in enumerate(linear_scores, start=1):
    print(f"Fold {i} accuracy: {score:.4f}")


print("Linear SVM 10-Fold Accuracy:")
print("Mean:", linear_scores.mean())
print("Std :", linear_scores.std())
```

```python
# RBF (Non-Linear) SVM Pipeline
rbf_pipeline = Pipeline([
    ("scaler", StandardScaler()),
    ("svm", SVC(kernel="rbf", C=1.0, gamma="scale"))
])

rbf_scores = cross_val_score(
    rbf_pipeline,
    X,
    y,
    cv=cv,
    scoring="accuracy"
)

for i, score in enumerate(rbf_scores, start=1):
    print(f"Fold {i} accuracy: {score:.4f}")


print("RBF SVM 10-Fold Accuracy:")
print("Mean:", rbf_scores.mean())
print("Std :", rbf_scores.std())
```

```python
# Compare Results
results = pd.DataFrame({
    "Model": ["Linear SVM", "RBF SVM"],
    "Mean Accuracy": [linear_scores.mean(), rbf_scores.mean()],
    "Std Dev": [linear_scores.std(), rbf_scores.std()]
})

results
```

## Example 2 - Oneclass SVM 
 # Anomalous Deliveries

<!-- Detect abnormal / risky deliveries:

    Extremely long delivery times
    Unusual pickup times 
    Route distance anomalies

No labels required (unsupervised learning) -->

```python
from sklearn.svm import OneClassSVM
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
```

```python
# Feature Selection for Anomaly Detection

# Important: One-Class SVM expects only “normal” operational features, not labels.

features = [
    "pickup_hour",
    "pickup_dayofweek",
    "distance",
    "delivery_duration_min"
]

X_ocsvm = df[features]
```

```python
# Scaling 

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_ocsvm)

# StandardScaler() standardizes features so that:
# Mean = 0
# Standard deviation = 1

# Mathematically, for each feature:

# 𝑥 = (𝑥 − 𝜇) / 𝜎

# where
#     μ = feature mean
#     σ = feature standard deviation

# Resulting values:

#     Can be negative
#     Can be greater than 1
#     Can be less than -1
#     No fixed bounds

# WHY SCALING IS REQUIRED: ?

# 1. SVM is distance-based
    # SVM uses dot products
    # Large-scale features dominate small-scale ones

# Example (logistics):
#     distance → 1–500 km
#     pickup_hour → 0–23

# Without scaling:
#     distance >>> pickup_hour
# So the model mostly learns distance only distance 

# 2. If features are not scaled:
    # Kernel values collapse to ~0
    # Decision boundary becomes meaningless

# 3. Faster and more stable optimization
#     Improves convergence
#     Reduces numerical instability


# StandardScaler standardizes each feature to zero mean and unit variance.
# This is critical for SVMs because they are distance-based models and are
# highly sensitive to feature scale, especially when using non-linear
# kernels such as RBF.
```

```python
# Train One-Class SVM
#     Key hyperparameters (from a logistics intuition):
#     nu: expected % of anomalies (5–10%)
#     gamma: non-linear boundary sensitivity

ocsvm = OneClassSVM(
    kernel="rbf",
    nu=0.05,        # assume ~5% abnormal deliveries
    gamma="scale"
)

ocsvm.fit(X_scaled)
```

```python
# Predict Anomalies
df["anomaly_flag"] = ocsvm.predict(X_scaled)


# Output meaning:

# +1 → Normal delivery
# -1 → Anomalous delivery
```

```python
df["anomaly_flag"].value_counts()
```

```python
# Convert to Binary Label (for our clarity)
```

```python
df["is_anomalous"] = (df["anomaly_flag"] == -1).astype(int)

df["is_anomalous"].value_counts()
```

```python
# Inspect Anomalous Deliveries

df[df["is_anomalous"] == 1][
    ["pickup_time", "delivery_time", "distance", "delivery_duration_min"]
].head(10)
```

```python
plt.figure(figsize=(7, 4))

plt.scatter(
    df["distance"],
    df["delivery_duration_min"],
    c=df["is_anomalous"],
    alpha=0.6
)

plt.xlabel("Distance")
plt.ylabel("Delivery Duration (min)")
plt.title("One-Class SVM: Anomalous Deliveries")
plt.show()
```
